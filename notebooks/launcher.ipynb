# tensorcas — Notebook Launcher

Clone the repo and run any benchmark notebook from here.

**Steps:**
1. Run the *Setup* cell to clone the repo and install core dependencies
2. Run the *install* cell for the notebook you want
3. Run the *Execute* cell to run it — output saves to `notebooks/<name>_out.ipynb`

> Notebooks can also be opened directly in Colab via the links below.

### GPT-2 fine-tuning benchmark

<a href="https://colab.research.google.com/github/Olamyy/tensorcas/blob/hash-cache-no-op-path/notebooks/gpt2_finetuning.ipynb" target="_parent"><img src="https://colab.research.google.com/assets/colab-badge.svg" alt="Open In Colab"/></a>

```python
!pip install -q transformers safetensors dvc
```

```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import sys

nb_path = 'notebooks/gpt2_finetuning_out.ipynb'
with open(nb_path) as f:
    nb = nbformat.read(f, as_version=4)

ep = ExecutePreprocessor(timeout=1800, kernel_name='python3')
ep.preprocess(nb, {'metadata': {'path': 'notebooks/'}})

# Print all cell outputs
for cell in nb.cells:
    if cell.cell_type != 'code':
        continue
    for output in cell.outputs:
        text = output.get('text', '') or output.get('data', {}).get('text/plain', '')
        if text:
            print(text, end='')
    sys.stdout.flush()

out_path = 'notebooks/gpt2_finetuning_out_out.ipynb'
with open(out_path, 'w') as f:
    nbformat.write(nb, f)
print('\nDone — output saved to ' + out_path)
```
