# tensorcas — Notebook Launcher

Clone the repo and run any benchmark notebook from here.

**Steps:**
1. Run the *Setup* cell to clone the repo and install core dependencies
2. Run the *install* cell for the notebook you want
3. Run the *Execute* cell to run it — output saves to `notebooks/<name>_out.ipynb`

> Notebooks can also be opened directly in Colab via the links below.

## Setup

Clone the repo and install `nbconvert` + core dependencies.

```python
!git clone -q --branch hash-cache-no-op-path https://github.com/Olamyy/tensorcas.git tensorcas
%cd tensorcas
!pip install -q nbconvert zstandard
import subprocess, sys
print('Repo cloned. Branch:', subprocess.check_output(['git', 'branch', '--show-current']).decode().strip())
```

Each section has:
- A Colab badge to open the notebook directly

```python
!jupyter nbconvert --to notebook --execute \
    --ExecutePreprocessor.timeout=1800 \
    --output notebooks/gpt2_finetuning_out.ipynb \
    notebooks/gpt2_finetuning.ipynb
print("Done — output: notebooks/gpt2_finetuning_out.ipynb")
```
